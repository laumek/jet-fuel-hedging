```python
import pandas as pd
from pathlib import Path

def load_jet_fuel(
    path: str | Path = "data/DJFUELUSGULF.csv",
    start: str = "2007-01-01"
) -> pd.Series:
    """
    Load daily US Gulf Coast jet fuel prices ($/gallon) from a FRED CSV.
    Returns a clean daily price Series indexed by date, starting at `start`.
    """
    data = pd.read_csv(path, index_col=0, parse_dates=True)

    data = data.loc[start:]
    print(data.isnull().sum())
    data = pd.to_numeric(data.iloc[:, 0], errors="coerce")
    data = data.dropna()
    data = data.rename("price")
    print(data.min(), data.idxmin(), data.max(), data.idxmax())

    return data
```

```python
prices = load_jet_fuel()
```

```python
dates = prices.index.to_series()
gaps = dates.diff()
gap_days = gaps.dt.days
big_gaps = gap_days[gap_days > 4]
print(f"Number of gaps over 4 days: {len(big_gaps)}")
print(big_gaps.sort_values(ascending=False).head(15))
print(big_gaps.groupby(big_gaps.index.year).size())
```
